# Regional margin check

A synthetic notebook written by hand for KumoSQL's tests. It loads orders, joins targets and plots margin by region.
It copies no real notebook.

```python
import pandas as pd
print('pandas', pd.__version__)
```

```python
orders = pd.read_gbq(
    "SELECT order_id, region, sales, profit FROM `acme-analytics.sales.orders`",
    project_id="acme-bi",
)
```

```python
by_region = (
    orders.groupby("region", as_index=False)
    .agg(sales=("sales", "sum"), profit=("profit", "sum"))
    .assign(margin=lambda d: (d.profit / d.sales).round(2))
)
by_region[["region", "margin"]]
```

```python
ax = by_region.plot.bar(x="region", y="margin", legend=False)
ax.axhline(0.25, color="#e15759")
```

```python
targets = pd.read_csv("targets.csv")
gap = by_region.merge(targets, on="region")["target"]
```
